```python
class Vehicle:                                 #Create a Vehicle class with maxspeed and color instance attributes.
    def __init__(self, maxspeed, color):
        self.maxspeed = maxspeed
        self.color = color

    def display_info(self):                    #Create new method which prints maxspeed and the color for the given vehicle.
        print(f"Vehicle Color: {self.color}")
        print(f"Vehicle Max Speed: {self.maxspeed} km/h")
```

```python
class Vehicle:                                       #Create a Train child class that inherits from the Vehicle class. The default fare of any vehicle is seating capacity * 35. If Vehicle is Train instance, we need to add an extra 10% on full fare as a maintenance charge.
    def __init__(self, maxspeed, seating_capacity):
        self.maxspeed = maxspeed
        self.seating_capacity = seating_capacity

    def fare(self):
        return self.seating_capacity * 35

class Train (Vehicle):
    def fare(self):
        base_fare = super().fare()
        total_fare = base_fare + (base_fare * 0.10)
        return total_fare
```
